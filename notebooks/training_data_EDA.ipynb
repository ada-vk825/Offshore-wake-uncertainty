```python
# Power Distribution
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Overall power distribution
axes[0,0].hist(df['power_kw'], bins=80, color='steelblue', edgecolor='white', alpha=0.8)
axes[0,0].axvline(df['power_kw'].mean(), color='red', linestyle='--', label=f'Mean={df.power_kw.mean():.0f} kW')
axes[0,0].axvline(df['power_kw'].median(), color='orange', linestyle='--', label=f'Median={df.power_kw.median():.0f} kW')
axes[0,0].set_xlabel('Power (kW)')
axes[0,0].set_ylabel('Count')
axes[0,0].set_title('Overall Power Distribution')
axes[0,0].legend()

# Power distribution per farm
for farm in df.farm_name.unique():
    farm_data = df[df.farm_name == farm]['power_kw']
    axes[0,1].hist(farm_data, bins=50, alpha=0.5, label=farm)
axes[0,1].set_xlabel('Power (kW)')
axes[0,1].set_ylabel('Count')
axes[0,1].set_title('Power Distribution by Farm')
axes[0,1].legend(fontsize=9)

# Power ratio (power / rated_power)
df['power_ratio'] = df['power_kw'] / df['target_rated_power_kw']
axes[1,0].hist(df['power_ratio'], bins=80, color='darkgreen', edgecolor='white', alpha=0.8)
axes[1,0].axvline(df['power_ratio'].mean(), color='red', linestyle='--', label=f'Mean={df.power_ratio.mean():.3f}')
axes[1,0].set_xlabel('Power Ratio (P / P_rated)')
axes[1,0].set_ylabel('Count')
axes[1,0].set_title('Normalised Power Ratio Distribution')
axes[1,0].legend()

# Box plot per farm
df.boxplot(column='power_ratio', by='farm_name', ax=axes[1,1])
axes[1,1].set_xlabel('Farm')
axes[1,1].set_ylabel('Power Ratio')
axes[1,1].set_title('Power Ratio by Farm')
plt.suptitle('')

plt.tight_layout()
plt.savefig('eda_01_power_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

print(f'\nPower ratio statistics:')
print(f'  Mean: {df.power_ratio.mean():.4f}')
print(f'  Std:  {df.power_ratio.std():.4f}')
print(f'  Turbines at >95% rated: {(df.power_ratio > 0.95).sum()} ({(df.power_ratio > 0.95).mean()*100:.1f}%)')
print(f'  Turbines at <50% rated: {(df.power_ratio < 0.50).sum()} ({(df.power_ratio < 0.50).mean()*100:.1f}%)')
```

```python
# Atmospheric & Scenario Features
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

# Wind speed distribution
axes[0].hist(df.groupby('scenario_id')['ws_free'].first(), bins=30, color='skyblue', edgecolor='white')
axes[0].set_xlabel('Wind Speed (m/s)')
axes[0].set_ylabel('Count (scenarios)')
axes[0].set_title('Wind Speed Distribution')

# Wind direction distribution
wd_counts = df.groupby('scenario_id')['wd'].first().value_counts().sort_index()
axes[1].bar(wd_counts.index, wd_counts.values, width=25, color='coral', edgecolor='white')
axes[1].set_xlabel('Wind Direction (deg)')
axes[1].set_ylabel('Count (scenarios)')
axes[1].set_title('Wind Direction Distribution')

# TI distribution
axes[2].hist(df.groupby('scenario_id')['ti'].first(), bins=20, color='mediumpurple', edgecolor='white')
axes[2].set_xlabel('Turbulence Intensity')
axes[2].set_ylabel('Count (scenarios)')
axes[2].set_title('TI Distribution')

plt.tight_layout()
plt.savefig('eda_02_atmospheric_features.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
FEATURES = [
    # Atmospheric conditions
    'ws_free', 'wd', 'ti',
    # Target turbine specs
    'target_rotor_dia_m', 'target_hub_height_m', 'target_rated_power_kw',
    # Neighbour farm specs
    'nb_distance_km', 'nb_direction_deg', 'nb_n_turbines', 'nb_spacing_D',
    'nb_rotor_dia_m', 'nb_hub_height_m', 'nb_rated_power_kw',
    # Physics-informed blocking features
    'pi_n_blocking', 'pi_blocking_ratio', 'pi_blocking_distance',
    'pi_streamwise_dist_1st', 'pi_lateral_dist_1st', 'pi_rotor_dia_1st',
]

TARGET = 'power_ratio'
```

```python

#TARGET FARM CONFIG
TARGET_CONFIG = {
    'n_turbines': 80,
    'diameter': 80,        # rotor diameter (m)
    'hub_height': 70,      # hub height (m)
    'rated_power': 2000,   # rated power (kW)
    'spacing_D': 7,        # spacing in rotor diameters
}

# NEIGHBOUR FARM CONFIG
NEIGHBOUR_CONFIG = {
    'n_turbines': 50,
    'diameter': 80,
    'hub_height': 70,
    'rated_power': 2000,
    'spacing_D': 7,
    'distance_km': 15,       # distance from target farm centre
    'direction_deg': 270,    # direction FROM target TO neighbour (270 = West)
}

# ATMOSPHERIC CONDITIONS
WIND_CONFIG = {
    'ws': 9.0,    # wind speed (m/s)
    'wd': 270.0,  # wind direction (degrees, where wind comes FROM)
    'ti': 0.06,   # turbulence intensity
}

# MONTE CARLO
N_MC = 100  # number of Monte Carlo iterations

# Jensen wake expansion coefficient (same as in utils.py)
KW_JENSEN = 0.04

print(f'Target farm: {TARGET_CONFIG["n_turbines"]} turbines, {TARGET_CONFIG["diameter"]}m rotor, {TARGET_CONFIG["rated_power"]} kW')
print(f'Neighbour farm: {NEIGHBOUR_CONFIG["n_turbines"]} turbines at {NEIGHBOUR_CONFIG["distance_km"]} km, {NEIGHBOUR_CONFIG["direction_deg"]}°')
print(f'Wind: {WIND_CONFIG["ws"]} m/s from {WIND_CONFIG["wd"]}°, TI={WIND_CONFIG["ti"]}')
print(f'Monte Carlo iterations: {N_MC}')
```

```python
from shapely.geometry import Polygon

turbines_per_side = int(np.ceil(np.sqrt(TARGET_CONFIG['n_turbines'])))
target_spacing_m = TARGET_CONFIG['spacing_D'] * TARGET_CONFIG['diameter']
target_side_length = (turbines_per_side + 1) * target_spacing_m

# Centre at origin
target_cx, target_cy = 0.0, 0.0
target_poly = Polygon([
    (target_cx - target_side_length/2, target_cy - target_side_length/2),
    (target_cx + target_side_length/2, target_cy - target_side_length/2),
    (target_cx + target_side_length/2, target_cy + target_side_length/2),
    (target_cx - target_side_length/2, target_cy + target_side_length/2),
])

print(f'Target polygon side length: {target_side_length:.0f} m')
```

```python
results = []

for mc_iter in range(N_MC):
    seed = mc_iter * 100  # different seed each iteration
  
    x_target, y_target, _ = place_turbines(
        target_poly, 
        TARGET_CONFIG['n_turbines'], 
        target_spacing_m, 
        seed=seed
    )
    n_placed_target = len(x_target)
    
    
    nb_poly, nb_cx, nb_cy = make_neighbour_polygon(
        target_cx, target_cy,
        NEIGHBOUR_CONFIG['distance_km'],
        NEIGHBOUR_CONFIG['direction_deg'],
        NEIGHBOUR_CONFIG['n_turbines'],
        NEIGHBOUR_CONFIG['spacing_D'],
        NEIGHBOUR_CONFIG['diameter']
    )
    nb_spacing_m = NEIGHBOUR_CONFIG['spacing_D'] * NEIGHBOUR_CONFIG['diameter']
    x_nb, y_nb, _ = place_turbines(
        nb_poly, 
        NEIGHBOUR_CONFIG['n_turbines'], 
        nb_spacing_m, 
        seed=seed + 1
    )
    n_placed_nb = len(x_nb)
    
    # This gives P_without_neighbour (only intra-farm wakes)
    site = UniformSite(p_wd=[1], ti=WIND_CONFIG['ti'])
    target_wt = GenericWindTurbine(
        'Target', 
        TARGET_CONFIG['diameter'],
        TARGET_CONFIG['hub_height'], 
        TARGET_CONFIG['rated_power']
    )
    wfm_alone = Nygaard_2022(site, target_wt)
    sim_alone = wfm_alone(x_target, y_target, wd=[WIND_CONFIG['wd']], ws=[WIND_CONFIG['ws']])
    power_alone_array = sim_alone.Power.values  # shape: (n_target, 1, 1) in Watts
    power_alone_kw = power_alone_array[:, 0, 0] / 1000.0  # per turbine in kW
    P_without_neighbour = power_alone_kw.sum()  # total farm power without neighbour
    
    
    # Using compute_geometric_features from utils.py
    # This computes features considering ALL other turbines (same farm + neighbour)
    x_all = np.concatenate([x_target, x_nb])
    y_all = np.concatenate([y_target, y_nb])
    D_all = np.array(
        [TARGET_CONFIG['diameter']] * n_placed_target +
        [NEIGHBOUR_CONFIG['diameter']] * n_placed_nb
    )
    
    # Build feature rows for each target turbine
    feature_rows = []
    for t_idx in range(n_placed_target):
        # All OTHER turbines (excluding self)
        other_mask = np.ones(len(x_all), dtype=bool)
        other_mask[t_idx] = False
        other_x = x_all[other_mask]
        other_y = y_all[other_mask]
        other_D = D_all[other_mask]
        
        # Compute geometric features (same function as training data generation)
        geo_feats = compute_geometric_features(
            x_all[t_idx], y_all[t_idx], D_all[t_idx],
            other_x, other_y, other_D,
            WIND_CONFIG['wd'], kw=KW_JENSEN
        )
        
        row = {
            'ws_free': WIND_CONFIG['ws'],
            'wd': WIND_CONFIG['wd'],
            'ti': WIND_CONFIG['ti'],
            'target_rotor_dia_m': TARGET_CONFIG['diameter'],
            'target_hub_height_m': TARGET_CONFIG['hub_height'],
            'target_rated_power_kw': TARGET_CONFIG['rated_power'],
            'nb_distance_km': NEIGHBOUR_CONFIG['distance_km'],
            'nb_direction_deg': NEIGHBOUR_CONFIG['direction_deg'],
            'nb_n_turbines': n_placed_nb,
            'nb_spacing_D': NEIGHBOUR_CONFIG['spacing_D'],
            'nb_rotor_dia_m': NEIGHBOUR_CONFIG['diameter'],
            'nb_hub_height_m': NEIGHBOUR_CONFIG['hub_height'],
            'nb_rated_power_kw': NEIGHBOUR_CONFIG['rated_power'],
            # PI features (from compute_geometric_features)
            'pi_n_blocking': geo_feats['pi_n_blocking'],
            'pi_blocking_ratio': geo_feats['pi_blocking_ratio'],
            'pi_blocking_distance': geo_feats['pi_blocking_distance'],
            'pi_streamwise_dist_1st': geo_feats['pi_streamwise_dist_1st'],
            'pi_lateral_dist_1st': geo_feats['pi_lateral_dist_1st'],
            'pi_rotor_dia_1st': geo_feats['pi_rotor_dia_1st'],
        }
        feature_rows.append(row)
    
   
    feature_df = pd.DataFrame(feature_rows)
    predicted_power_ratio = best_model.predict(feature_df[FEATURES])
    predicted_power_kw = predicted_power_ratio * TARGET_CONFIG['rated_power']
    P_with_neighbour = predicted_power_kw.sum()  # total farm power with neighbour
    
    if P_without_neighbour > 0:
        wake_loss_pct = (1 - P_with_neighbour / P_without_neighbour) * 100
    else:
        wake_loss_pct = 0.0
    
    results.append({
        'mc_iter': mc_iter,
        'n_target_placed': n_placed_target,
        'n_nb_placed': n_placed_nb,
        'P_without_neighbour_kw': P_without_neighbour,
        'P_with_neighbour_kw': P_with_neighbour,
        'wake_loss_pct': wake_loss_pct,
    })
    
    if (mc_iter + 1) % 10 == 0:
        print(f'  Iteration {mc_iter+1}/{N_MC} | Wake loss = {wake_loss_pct:.2f}% | '
              f'P_alone={P_without_neighbour:.0f} kW, P_with_nb={P_with_neighbour:.0f} kW')

df_results = pd.DataFrame(results)
print(f'\nMonte Carlo complete: {N_MC} iterations.')
```

```python
def run_monte_carlo_wake_loss(target_config, neighbour_config, wind_config, n_mc=30):
    """
    Run Monte Carlo wake loss simulation.
    
    Returns:
        dict with 'mean', 'std', 'ci_low', 'ci_high', 'all_values'
    """
    # Create target polygon
    turbines_per_side = int(np.ceil(np.sqrt(target_config['n_turbines'])))
    target_spacing_m = target_config['spacing_D'] * target_config['diameter']
    target_side_length = (turbines_per_side + 1) * target_spacing_m
    target_cx, target_cy = 0.0, 0.0
    
    target_poly = Polygon([
        (target_cx - target_side_length/2, target_cy - target_side_length/2),
        (target_cx + target_side_length/2, target_cy - target_side_length/2),
        (target_cx + target_side_length/2, target_cy + target_side_length/2),
        (target_cx - target_side_length/2, target_cy + target_side_length/2),
    ])
    
    nb_spacing_m = neighbour_config['spacing_D'] * neighbour_config['diameter']
    wake_losses = []
    
    for mc_iter in range(n_mc):
        seed = mc_iter * 100
        
        
        x_target, y_target, _ = place_turbines(
            target_poly, target_config['n_turbines'], target_spacing_m, seed=seed
        )
        n_placed_target = len(x_target)
        
    
        nb_poly, nb_cx, nb_cy = make_neighbour_polygon(
            target_cx, target_cy,
            neighbour_config['distance_km'],
            neighbour_config['direction_deg'],
            neighbour_config['n_turbines'],
            neighbour_config['spacing_D'],
            neighbour_config['diameter']
        )
        x_nb, y_nb, _ = place_turbines(
            nb_poly, neighbour_config['n_turbines'], nb_spacing_m, seed=seed + 1
        )
        n_placed_nb = len(x_nb)
        
        
        site = UniformSite(p_wd=[1], ti=wind_config['ti'])
        target_wt = GenericWindTurbine(
            'Target', target_config['diameter'],
            target_config['hub_height'], target_config['rated_power']
        )
        wfm_alone = Nygaard_2022(site, target_wt)
        sim_alone = wfm_alone(x_target, y_target, wd=[wind_config['wd']], ws=[wind_config['ws']])
        power_alone_kw = sim_alone.Power.values[:, 0, 0] / 1000.0
        P_without_neighbour = power_alone_kw.sum()
 
        x_all = np.concatenate([x_target, x_nb])
        y_all = np.concatenate([y_target, y_nb])
        D_all = np.array(
            [target_config['diameter']] * n_placed_target +
            [neighbour_config['diameter']] * n_placed_nb
        )
        
        feature_rows = []
        for t_idx in range(n_placed_target):
            other_mask = np.ones(len(x_all), dtype=bool)
            other_mask[t_idx] = False
            other_x = x_all[other_mask]
            other_y = y_all[other_mask]
            other_D = D_all[other_mask]
            
            geo_feats = compute_geometric_features(
                x_all[t_idx], y_all[t_idx], D_all[t_idx],
                other_x, other_y, other_D,
                wind_config['wd'], kw=KW_JENSEN
            )
            
            row = {
                'ws_free': wind_config['ws'],
                'wd': wind_config['wd'],
                'ti': wind_config['ti'],
                'target_rotor_dia_m': target_config['diameter'],
                'target_hub_height_m': target_config['hub_height'],
                'target_rated_power_kw': target_config['rated_power'],
                'nb_distance_km': neighbour_config['distance_km'],
                'nb_direction_deg': neighbour_config['direction_deg'],
                'nb_n_turbines': n_placed_nb,
                'nb_spacing_D': neighbour_config['spacing_D'],
                'nb_rotor_dia_m': neighbour_config['diameter'],
                'nb_hub_height_m': neighbour_config['hub_height'],
                'nb_rated_power_kw': neighbour_config['rated_power'],
                'pi_n_blocking': geo_feats['pi_n_blocking'],
                'pi_blocking_ratio': geo_feats['pi_blocking_ratio'],
                'pi_blocking_distance': geo_feats['pi_blocking_distance'],
                'pi_streamwise_dist_1st': geo_feats['pi_streamwise_dist_1st'],
                'pi_lateral_dist_1st': geo_feats['pi_lateral_dist_1st'],
                'pi_rotor_dia_1st': geo_feats['pi_rotor_dia_1st'],
            }
            feature_rows.append(row)
        
        feature_df = pd.DataFrame(feature_rows)
        predicted_power_ratio = best_model.predict(feature_df[FEATURES])
        predicted_power_kw = predicted_power_ratio * target_config['rated_power']
        P_with_neighbour = predicted_power_kw.sum()
    
        if P_without_neighbour > 0:
            wake_loss_pct = (1 - P_with_neighbour / P_without_neighbour) * 100
        else:
            wake_loss_pct = 0.0
        wake_loss_pct = max(0, wake_loss_pct)
        wake_losses.append(wake_loss_pct)
    
    wake_losses = np.array(wake_losses)
    return {
        'mean': wake_losses.mean(),
        'std': wake_losses.std(),
        'ci_low': np.percentile(wake_losses, 2.5),
        'ci_high': np.percentile(wake_losses, 97.5),
        'all_values': wake_losses,
    }
```

```python
BASE_TARGET = {
    'n_turbines': 80,
    'diameter': 80,
    'hub_height': 70,
    'rated_power': 2000,
    'spacing_D': 7,
}

BASE_NEIGHBOUR = {
    'n_turbines': 50,
    'diameter': 80,
    'hub_height': 70,
    'rated_power': 2000,
    'spacing_D': 7,
    'distance_km': 15,
    'direction_deg': 270,
}

BASE_WIND = {
    'ws': 9.0,
    'wd': 270.0,
    'ti': 0.06,
}

# MC iterations per parameter value (lower = faster, 30 is good enough for trends)
N_MC_SENSITIVITY = 30

print('Base configuration set.')
```

```python


distances = [5, 10, 15, 20, 25, 30]
distance_results = []

print('Running sensitivity: DISTANCE')
for dist in distances:
    nb_config = BASE_NEIGHBOUR.copy()
    nb_config['distance_km'] = dist
    
    result = run_monte_carlo_wake_loss(BASE_TARGET, nb_config, BASE_WIND, n_mc=N_MC_SENSITIVITY)
    distance_results.append(result)
    print(f'  Distance = {dist} km → Wake loss = {result["mean"]:.2f}% ± {result["std"]:.2f}%')
```

```python
wind_directions = np.arange(0, 360, 10)
wd_results = []

print('Running sensitivity: WIND DIRECTION')
for wd in wind_directions:
    wind_config = BASE_WIND.copy()
    wind_config['wd'] = wd
    
    result = run_monte_carlo_wake_loss(BASE_TARGET, BASE_NEIGHBOUR, wind_config, n_mc=N_MC_SENSITIVITY)
    wd_results.append(result)
    print(f'  Wind direction = {wd}° → Wake loss = {result["mean"]:.2f}% ± {result["std"]:.2f}%')
```

```python
wind_speeds = [5, 7, 9, 11, 13]
ws_results = []

print('Running sensitivity: WIND SPEED')
for ws in wind_speeds:
    wind_config = BASE_WIND.copy()
    wind_config['ws'] = ws
    
    result = run_monte_carlo_wake_loss(BASE_TARGET, BASE_NEIGHBOUR, wind_config, n_mc=N_MC_SENSITIVITY)
    ws_results.append(result)
    print(f'  Wind speed = {ws} m/s → Wake loss = {result["mean"]:.2f}% ± {result["std"]:.2f}%')
```
